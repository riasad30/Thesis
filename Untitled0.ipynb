```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# OUTLIER & NUMERICAL DATA QUALITY ANALYSIS
# ============================================================

numeric_cols = [
    'LIMIT_BAL',
    'AGE',
    'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3',
    'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6',
    'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
    'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6'
]

outlier_results = []

for col in numeric_cols:

    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (clean_df[col] < lower_bound) |
        (clean_df[col] > upper_bound)
    )

    outlier_count = outlier_mask.sum()
    outlier_percentage = (outlier_count / len(clean_df)) * 100

    outlier_results.append({
        'Variable': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outlier_count,
        'Outlier %': round(outlier_percentage, 2),
        'Minimum': clean_df[col].min(),
        'Maximum': clean_df[col].max(),
        'Mean': clean_df[col].mean(),
        'Median': clean_df[col].median(),
        'Skewness': clean_df[col].skew()
    })

outlier_df = pd.DataFrame(outlier_results)

print("=" * 80)
print("OUTLIER ANALYSIS")
print("=" * 80)

print(outlier_df.to_string(index=False))

# Save report
outlier_df.to_csv("outlier_analysis_report.csv", index=False)

print("\nOutlier analysis report saved as:")
print("outlier_analysis_report.csv")
```

One-hot encode + scale

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# ============================================================
# PREPROCESSING PIPELINE
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numerical_features
        ),
        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply learned transformations to test data
X_test_processed = preprocessor.transform(X_test)

print("Original training shape:", X_train.shape)
print("Original testing shape :", X_test.shape)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
```

A. Chi-square test

```python
from scipy.stats import chi2_contingency


cat_features = [
'SEX',
'EDUCATION',
'MARRIAGE',
'PAY_0',
'PAY_2',
'PAY_3',
'PAY_4',
'PAY_5',
'PAY_6'
]


chi_results=[]

for col in cat_features:

    table=pd.crosstab(
        clean_df[col],
        y
    )

    chi2,p,_,_=chi2_contingency(table)

    chi_results.append(
        [col,chi2,p]
    )


chi_df=pd.DataFrame(
    chi_results,
    columns=[
        "Feature",
        "Chi-square",
        "p-value"
    ]
)

chi_df.sort_values(
    "p-value"
)
```

B. Mann–Whitney U Test

```python
from scipy.stats import mannwhitneyu


num_features=[
'LIMIT_BAL',
'AGE',
'BILL_AMT1',
'BILL_AMT2',
'BILL_AMT3',
'BILL_AMT4',
'BILL_AMT5',
'BILL_AMT6',
'PAY_AMT1',
'PAY_AMT2',
'PAY_AMT3',
'PAY_AMT4',
'PAY_AMT5',
'PAY_AMT6'
]


mw_results=[]


for col in num_features:

    group0 = clean_df[
        clean_df[target]==0
    ][col]

    group1 = clean_df[
        clean_df[target]==1
    ][col]


    stat,p=mannwhitneyu(
        group0,
        group1
    )


    mw_results.append(
        [col,stat,p]
    )


mw_df=pd.DataFrame(
    mw_results,
    columns=[
        "Feature",
        "U-statistic",
        "p-value"
    ]
)


mw_df.sort_values("p-value")
```

e. Credit Utilization Ratio

```python
clean_df['CREDIT_UTILIZATION'] = (
    clean_df['AVG_BILL_AMT'] /
    clean_df['LIMIT_BAL']
)
```

f. Payment Delay Score

```python
pay_cols=[
'PAY_0',
'PAY_2',
'PAY_3',
'PAY_4',
'PAY_5',
'PAY_6'
]

clean_df['AVG_PAYMENT_DELAY'] = (
    clean_df[pay_cols].mean(axis=1)
)
```

```python
clean_df['REMAINING_DEBT'] = (
    clean_df['TOTAL_BILL_AMT']
    -
    clean_df['TOTAL_PAY_AMT']
)
```

```python
util_cols = [
    'BILL_AMT1',
    'BILL_AMT2',
    'BILL_AMT3',
    'BILL_AMT4',
    'BILL_AMT5',
    'BILL_AMT6'
]


clean_df['MAX_BILL_AMT'] = (
    clean_df[util_cols].max(axis=1)
)
```

```python
clean_df['PAYMENT_RATIO'] = (
    clean_df['AVG_PAY_AMT']
    /
    (clean_df['AVG_BILL_AMT']+1)
)
```

```python
clean_df['TOTAL_DELAY_MONTHS'] = (
    clean_df[pay_cols] > 0
).sum(axis=1)
```

```python
clean_df['MAX_PAYMENT_DELAY'] = (
    clean_df[pay_cols].max(axis=1)
)
```

```python
clean_df['BILL_TREND'] = (
    clean_df['BILL_AMT1']
    -
    clean_df['BILL_AMT6']
)
```

```python
# ==========================================
# Advanced Payment Behavior Features
# ==========================================

pay_amt_cols = [
    'PAY_AMT1',
    'PAY_AMT2',
    'PAY_AMT3',
    'PAY_AMT4',
    'PAY_AMT5',
    'PAY_AMT6'
]


# 1. Total payment across 6 months
clean_df['TOTAL_PAYMENT'] = (
    clean_df[pay_amt_cols]
    .sum(axis=1)
)


# 2. Average monthly payment
clean_df['AVERAGE_PAYMENT'] = (
    clean_df[pay_amt_cols]
    .mean(axis=1)
)


# 3. Maximum payment made in a month
clean_df['MAX_PAYMENT'] = (
    clean_df[pay_amt_cols]
    .max(axis=1)
)


# 4. Minimum payment made in a month
clean_df['MIN_PAYMENT'] = (
    clean_df[pay_amt_cols]
    .min(axis=1)
)


# 5. Payment consistency / stability
clean_df['PAYMENT_STD'] = (
    clean_df[pay_amt_cols]
    .std(axis=1)
)


# 6. Number of months with zero payment
clean_df['ZERO_PAYMENT_MONTHS'] = (
    clean_df[pay_amt_cols] == 0
).sum(axis=1)
```

```python
import optuna

from catboost import CatBoostClassifier

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    f1_score
)
```

```python
def objective(trial):

    params = {

        "iterations": trial.suggest_int(
            "iterations",
            500,
            2000
        ),

        "depth": trial.suggest_int(
            "depth",
            4,
            10
        ),

        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.01,
            0.1,
            log=True
        ),

        "l2_leaf_reg": trial.suggest_float(
            "l2_leaf_reg",
            1,
            10
        ),

        "random_strength": trial.suggest_float(
            "random_strength",
            0,
            2
        ),

        "bagging_temperature": trial.suggest_float(
            "bagging_temperature",
            0,
            5
        ),

        "loss_function": "Logloss",

        "eval_metric": "AUC",

        "random_seed": 42,

        "verbose": 0
    }


    model = CatBoostClassifier(**params)


    model.fit(
        X_train,
        y_train
    )


    pred_prob = model.predict_proba(
        X_test
    )[:,1]


    auc = roc_auc_score(
        y_test,
        pred_prob
    )


    return auc
```

```python
study = optuna.create_study(
    direction="maximize"
)


study.optimize(
    objective,
    n_trials=50
)
```

```python
print("Best ROC-AUC:")
print(study.best_value)


print("\nBest Parameters:")
print(study.best_params)
```

```python
from catboost import CatBoostClassifier

# Instantiate the final model using the best hyperparameters found by Optuna
final_catboost = CatBoostClassifier(
    **study.best_params,
    loss_function="Logloss",
    eval_metric="AUC",
    random_seed=42,
    verbose=0
)

# Train the final model
final_catboost.fit(X_train, y_train)

# Make predictions
pred = final_catboost.predict(
    X_test
)

prob = final_catboost.predict_proba(
    X_test
)[:,1]

# Evaluate performance
print("Accuracy:",
      accuracy_score(y_test,pred))

print("F1:",
      f1_score(y_test,pred))

print("ROC-AUC:",
      roc_auc_score(y_test,prob))
```
