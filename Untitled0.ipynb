```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# OUTLIER & NUMERICAL DATA QUALITY ANALYSIS
# ============================================================

numeric_cols = [
    'LIMIT_BAL',
    'AGE',
    'BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3',
    'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6',
    'PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
    'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6'
]

outlier_results = []

for col in numeric_cols:

    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_mask = (
        (clean_df[col] < lower_bound) |
        (clean_df[col] > upper_bound)
    )

    outlier_count = outlier_mask.sum()
    outlier_percentage = (outlier_count / len(clean_df)) * 100

    outlier_results.append({
        'Variable': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': outlier_count,
        'Outlier %': round(outlier_percentage, 2),
        'Minimum': clean_df[col].min(),
        'Maximum': clean_df[col].max(),
        'Mean': clean_df[col].mean(),
        'Median': clean_df[col].median(),
        'Skewness': clean_df[col].skew()
    })

outlier_df = pd.DataFrame(outlier_results)

print("=" * 80)
print("OUTLIER ANALYSIS")
print("=" * 80)

print(outlier_df.to_string(index=False))

# Save report
outlier_df.to_csv("outlier_analysis_report.csv", index=False)

print("\nOutlier analysis report saved as:")
print("outlier_analysis_report.csv")
```

One-hot encode + scale

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# ============================================================
# PREPROCESSING PIPELINE
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            'num',
            StandardScaler(),
            numerical_features
        ),
        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_features
        )
    ]
)

# Fit ONLY on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply learned transformations to test data
X_test_processed = preprocessor.transform(X_test)

print("Original training shape:", X_train.shape)
print("Original testing shape :", X_test.shape)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
```

```python
# ============================================================
# CHI-SQUARE TEST
# ============================================================


categorical_cols = [

'SEX',
'EDUCATION',
'MARRIAGE',
'PAY_0',
'PAY_2',
'PAY_3',
'PAY_4',
'PAY_5',
'PAY_6'

]


chi_results=[]


for col in categorical_cols:


    table = pd.crosstab(
        clean_df[col],
        clean_df[target_col]
    )


    chi2,p,dof,expected = chi2_contingency(table)


    chi_results.append(
        [
            col,
            chi2,
            p
        ]
    )


chi_table = pd.DataFrame(
    chi_results,
    columns=[
        "Variable",
        "Chi-square",
        "p-value"
    ]
)


display(
    chi_table.round(4)
)
```

# **STEP 4: Feature Engineering**

```python
# ==========================================
# STEP 4.1: PAYMENT HISTORY FEATURES
# ==========================================

pay_cols = [
    'PAY_0',
    'PAY_2',
    'PAY_3',
    'PAY_4',
    'PAY_5',
    'PAY_6'
]


# Total delayed payment history

df['TOTAL_PAY_DELAY'] = df[pay_cols].sum(axis=1)


# Maximum delay

df['MAX_PAY_DELAY'] = df[pay_cols].max(axis=1)


# Average delay

df['AVG_PAY_DELAY'] = df[pay_cols].mean(axis=1)


# Number of months with delay

df['NUM_DELAY_MONTHS'] = (df[pay_cols] > 0).sum(axis=1)


print(df.shape)
```

```python
# ==========================================
# CREDIT UTILIZATION FEATURES
# ==========================================


bill_cols = [
    'BILL_AMT1',
    'BILL_AMT2',
    'BILL_AMT3',
    'BILL_AMT4',
    'BILL_AMT5',
    'BILL_AMT6'
]


for i, col in enumerate(bill_cols, start=1):

    df[f'CREDIT_UTIL_{i}'] = (
        df[col] /
        (df['LIMIT_BAL'] + 1)
    )


print(df.shape)
```

```python
# ==========================================
# FINANCIAL SUMMARY FEATURES
# ==========================================


payment_cols = [
    'PAY_AMT1',
    'PAY_AMT2',
    'PAY_AMT3',
    'PAY_AMT4',
    'PAY_AMT5',
    'PAY_AMT6'
]


# Total bill

df['TOTAL_BILL'] = (
    df[bill_cols].sum(axis=1)
)


# Total payment

df['TOTAL_PAYMENT'] = (
    df[payment_cols].sum(axis=1)
)


# Payment ability ratio

df['PAYMENT_BILL_RATIO'] = (
    df['TOTAL_PAYMENT'] /
    (df['TOTAL_BILL'] + 1)
)


print(df.shape)
```
