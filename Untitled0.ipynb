Investigate the 35 duplicates

```python
# Exact duplicate rows after removing ID
exact_duplicates = clean_df[clean_df.duplicated(keep=False)]

print("Number of rows involved in exact duplicates:",
      len(exact_duplicates))

print("\nNumber of duplicate groups:")

print(
    exact_duplicates
    .groupby(list(clean_df.columns))
    .size()
    .sort_values(ascending=False)
    .head(20)
)
```

```python
target_col = 'default payment next month'

feature_cols = [
    col for col in clean_df.columns
    if col != target_col
]

predictor_duplicates = (
    clean_df[
        clean_df.duplicated(
            subset=feature_cols,
            keep=False
        )
    ]
)

target_consistency = (
    predictor_duplicates
    .groupby(feature_cols)[target_col]
    .nunique()
)

print(
    "Repeated predictor profiles with different target values:",
    (target_consistency > 1).sum()
)
```

```python
target_col = 'default payment next month'

feature_cols = [
    col for col in clean_df.columns
    if col != target_col
]

repeated_profiles = clean_df[
    clean_df.duplicated(subset=feature_cols, keep=False)
].copy()

profile_summary = (
    repeated_profiles
    .groupby(feature_cols, dropna=False)
    .agg(
        number_of_records=(target_col, 'size'),
        number_of_target_classes=(target_col, 'nunique')
    )
    .reset_index()
)

print("Repeated predictor profiles:", len(profile_summary))

print("\nProfiles with different target values:")
print(
    profile_summary[
        profile_summary['number_of_target_classes'] > 1
    ].shape[0]
)

print("\nTotal observations involved in repeated profiles:",
      len(repeated_profiles))
```
